```python
# 1. Arithmetic operator :  + , - , * , / , ** %,
```

```python
x1 = 12
x2 = 10
```

```python
print("Addition:", x1+x2)
print("subtraction:", x1-x2)
print("multiplication:", x1*x2)
print("division:", x1/x2)
print("integer division:", x1//x2)
print("Reminder:", x1 % x2)
print("power:", x1 ** 2)
```

```python
# 	2. Comparison operator:  > , < , >=, <=, ==,!=
print("is x1 is greater than x2?:", x1 > x2)
print("is x1 is lesser than x2?:", x1 < x2)
print("is x1 and x2 are same?:", x1 == x2)
print("is x1 and x2 are not same?:", x1 != x2)
```
