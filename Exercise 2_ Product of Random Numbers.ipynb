```python
#Exercise 2: Product of Random Numbers
Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.
```

```python
import random      # Import Python's built-in random module for generating random numbers
import sys         # Import sys to access system-specific parameters like max integer size

def generate_two_random_numbers():
    """
    Generates two random integers between -sys.maxsize and sys.maxsize.
    """
    # Generate the first random integer in the full system integer range
    num1 = random.randint(-sys.maxsize, sys.maxsize)

    # Generate the second random integer in the same range
    num2 = random.randint(-sys.maxsize, sys.maxsize)

    # Return both numbers as a tuple
    return num1, num2

if __name__ == "__main__":
    # Call the function to get two random numbers
    n1, n2 = generate_two_random_numbers()

    # Print the generated numbers
    print(f"Random numbers generated: {n1} and {n2}")

    if n1 * n2 == int(input(f"What is {n1} times {n2}? ")):
        print("You got it right!")
    else:
        print("Sorry, you got it wrong.")
```
