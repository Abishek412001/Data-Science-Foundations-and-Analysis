```python
 # program to check whether a given number is prime or not
#A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. In other words, it can’t be evenly divided by any other number except 1 and itself
```

```python
a = int(input("Enter a number: "))
if a /a == 1 and a % a >=2:
        print(a, "is not a prime number")
else:
        print(a, "is a prime number")
```
