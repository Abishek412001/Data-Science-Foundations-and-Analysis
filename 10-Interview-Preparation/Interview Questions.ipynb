## 1. Difference Between Precision and Recall

Precision and Recall are evaluation metrics used in classification problems, especially when dealing with imbalanced datasets.

### Precision

Precision measures how many of the predicted positive cases are actually positive.

\text{Precision} = \frac{TP}{TP + FP}

Where:

* **TP** = True Positives
* **FP** = False Positives

### Meaning

* High precision means the model makes **fewer false positive errors**.
* Useful when the cost of false alarms is high.

### Example

In email spam detection:

* If the model marks 100 emails as spam,
* and 90 are actually spam,
* then precision = 90/100 = 0.90.

---

### Recall

Recall measures how many actual positive cases the model correctly identified.

\text{Recall} = \frac{TP}{TP + FN}

Where:

* **FN** = False Negatives

### Meaning

* High recall means the model misses very few positive cases.
* Useful when missing positive cases is costly.

### Example

In disease detection:

* If 100 people actually have a disease,
* and the model identifies 95 of them,
* then recall = 95/100 = 0.95.

---

### Key Difference

| Metric    | Focus                               | Important When             |
| --------- | ----------------------------------- | -------------------------- |
| Precision | Correctness of positive predictions | False positives are costly |
| Recall    | Finding all positive cases          | False negatives are costly |

### Simple Analogy

* **Precision** → “Out of all predicted positives, how many were correct?”
* **Recall** → “Out of all actual positives, how many did we find?”

## 2. What is Cross-Validation and Why is it Important?

### Definition

Cross-validation is a technique used to evaluate how well a machine learning model generalizes to unseen data.

Instead of training and testing the model only once, the dataset is split multiple times to ensure reliable evaluation.

---

### Most Common Method: K-Fold Cross-Validation

In K-Fold Cross-Validation:

1. The dataset is divided into **K equal parts (folds)**.
2. The model is trained on **K−1 folds**.
3. The remaining fold is used for testing.
4. This process repeats K times.
5. Final performance is the average of all runs.

Example with K = 5:

* Train on 4 folds
* Test on 1 fold
* Repeat 5 times

---

### Why Cross-Validation is Important in Binary Classification

#### 1. Better Performance Estimation

It provides a more reliable estimate of model accuracy than a single train-test split.

#### 2. Reduces Overfitting Risk

The model is tested on different subsets, helping detect whether it memorizes training data.

#### 3. Efficient Use of Data

Every data point gets a chance to be in both training and testing sets.

#### 4. Important for Small Datasets

When data is limited, cross-validation helps maximize training and evaluation efficiency.

#### 5. Helps Model Selection

Useful for comparing algorithms and tuning hyperparameters.

---

### Example

Suppose you are building a loan approval classifier:

* With only one train-test split, results may depend on random data selection.
* Cross-validation ensures the model performs consistently across different data samples.

---

### Conclusion

* **Precision** focuses on prediction quality.
* **Recall** focuses on capturing all positive cases.
* **Cross-validation** improves the reliability and robustness of binary classification models by testing them on multiple data splits.

## 1. What is Normalization & Standardization and How is it Helpful?

Normalization and Standardization are feature scaling techniques used in machine learning and statistics to bring data into a comparable range.

---

# Normalization

Normalization rescales data to a fixed range, usually between 0 and 1.

X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}

### Characteristics

* Values lie between **0 and 1**
* Preserves relationships among data points
* Sensitive to outliers

### Example

If marks range from 20 to 100:

* 20 becomes 0
* 100 becomes 1
* Other values fall between 0 and 1

---

# Standardization

Standardization transforms data so it has:

* Mean = 0
* Standard Deviation = 1

| Outlier Sensitivity | High                 | Lower                |
| Common Usage        | Neural Networks, KNN | Regression, SVM, PCA |

---

# Why Are They Helpful?

## 1. Improves Model Performance

Some algorithms perform better when features are on similar scales.

## 2. Faster Convergence

Gradient-based algorithms train faster with scaled data.

## 3. Prevents Dominance of Large Values

Without scaling, features with larger values may dominate smaller ones.

## 4. Important for Distance-Based Algorithms

Algorithms like:

* KNN
* K-Means
* SVM
* PCA

depend heavily on feature scale.

---

# Example

Suppose a dataset contains:

* Age: 20–60
* Salary: 20,000–2,00,000

Without scaling:

* Salary dominates calculations.

After scaling:

* Both features contribute fairly.

# 2. What Techniques Can Be Used to Address Multicollinearity in Multiple Linear Regression?

## What is Multicollinearity?

Multicollinearity occurs when independent variables are highly correlated with each other.

This causes:

* Unstable coefficient estimates
* Difficulty interpreting variables
* Reduced model reliability

---

# Techniques to Handle Multicollinearity

## 1. Remove Highly Correlated Features

Use:

* Correlation matrix
* Heatmaps

to identify strongly correlated variables and remove one of them.

### Example

If:

* House size
* Number of rooms

are highly correlated, remove one feature.

---

## 2. Variance Inflation Factor (VIF)

VIF measures how much multicollinearity exists.

VIF = \frac{1}{1 - R^2}

### Interpretation

| VIF Value | Meaning                   |
| --------- | ------------------------- |
| 1         | No multicollinearity      |
| 1–5       | Moderate                  |
| >10       | Serious multicollinearity |

Features with high VIF can be removed.

---

## 3. Principal Component Analysis (PCA)

Principal Component Analysis transforms correlated variables into uncorrelated principal components.

### Benefit

* Removes multicollinearity
* Reduces dimensionality

---

## 4. Ridge Regression

Ridge Regression adds a penalty term to reduce coefficient magnitude.

\text{Cost Function} = RSS + \lambda \sum \beta^2

### Benefit

* Handles multicollinearity effectively
* Prevents large coefficient values

---

## 5. Lasso Regression

Lasso Regression can shrink some coefficients to zero.

### Benefit

* Performs feature selection automatically
* Reduces multicollinearity

---

## 6. Collect More Data

Sometimes multicollinearity occurs because the dataset is too small.
Adding more observations can reduce the issue.

---

# Conclusion

* **Normalization** scales data to a fixed range.
* **Standardization** converts data to zero mean and unit variance.
* Both improve machine learning model performance.

To address **multicollinearity**:

* Remove correlated features
* Use VIF analysis
* Apply PCA
* Use Ridge/Lasso Regression
* Increase dataset size when possible.
