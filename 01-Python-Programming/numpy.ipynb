```python
# how we access the coloumn
#      row,columns
#      start:end-1 , start:end-1
data[:,0:1]  # accessing column
```

```python
data[0:1,:]  # accessing row
```

```python
print("Mean: " , np.mean(data[:,0:1]))
print("Median: " , np.median(data[:,0:1]))
print("standard deviation: " , np.std(data[:,0:1]))
print("standard deviation with rounding: " , np.round(np.std(data[:,0:1]),2))
print("variance: " , np.var(data[:,0:1]))
print("Min: " , np.min(data[:,0:1]))
print("Max: " , np.max(data[:,0:1]))
print("25th percentile: " , np.percentile(data[:,0:1],25))
print("50th percentile: " , np.percentile(data[:,0:1],50))
print("75th percentile: " , np.percentile(data[:,0:1],75))
```
