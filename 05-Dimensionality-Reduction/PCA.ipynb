```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("wine.csv")

# Display first 5 rows
print(df.head())

# Shape of dataset
print("Shape of Dataset:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Basic statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Plot histograms for all numerical columns
df.hist(figsize=(15,10), bins=20)
plt.tight_layout()
plt.show()

# Boxplots for numerical features
plt.figure(figsize=(15,8))

for i, column in enumerate(df.select_dtypes(include=np.number).columns):
    plt.subplot(4, 4, i+1) # Changed from (3,4) to (4,4) to accommodate all 14 columns
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()

# Density plots
df.select_dtypes(include=np.number).plot(
    kind='density',
    subplots=True,
    layout=(4,4), # Changed from (3,4) to (4,4) to accommodate all 14 columns
    figsize=(15,10),
    sharex=False
)

plt.tight_layout()
plt.show()

# Compute correlation matrix
correlation_matrix = df.corr(numeric_only=True)

# Display matrix
print(correlation_matrix)

# Find highly correlated features
threshold = 0.7

high_corr = correlation_matrix[
    (correlation_matrix > threshold) |
    (correlation_matrix < -threshold)
]

print(high_corr)

sns.pairplot(df.select_dtypes(include=np.number))
plt.show()
```

```python
# Pre-processing

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Example: removing target column
# Replace 'target' with actual target column name if available

X = df.drop('Type', axis=1) # Changed 'target' to 'Type'

# If dataset has only features:
# X = df


# Standardize data
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame (optional)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print(X_scaled.head())

# Apply PCA
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

# Explained variance ratio
explained_variance = pca.explained_variance_ratio_

print(explained_variance)

# Scree Plot
plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(explained_variance)+1),
    explained_variance,
    marker='o',
    linestyle='--'
)

plt.xlabel("Principal Components")
plt.ylabel("Explained Variance Ratio")
plt.title("Scree Plot")
plt.grid(True)

plt.show()

# Cumulative explained variance
cumulative_variance = np.cumsum(explained_variance)

print(cumulative_variance)

plt.figure(figsize=(8,5))

plt.plot(
    range(1, len(cumulative_variance)+1),
    cumulative_variance,
    marker='o'
)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Variance Plot")

plt.axhline(y=0.95, color='r', linestyle='--')

plt.grid(True)
plt.show()

# PCA with 95% variance retention
pca_final = PCA(n_components=0.95)

X_pca_final = pca_final.fit_transform(X_scaled)

print("Original Shape:", X_scaled.shape)
print("Reduced Shape:", X_pca_final.shape)

# Create DataFrame for principal components
pca_columns = [
    f'PC{i+1}'
    for i in range(X_pca_final.shape[1])
]

pca_df = pd.DataFrame(
    X_pca_final,
    columns=pca_columns
)

print(pca_df.head())

# Scatter plot of first two PCs

plt.figure(figsize=(8,6))

plt.scatter(
    pca_df['PC1'],
    pca_df['PC2']
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection")

plt.grid(True)
plt.show()

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Elbow Method
wcss = []

for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=42
    )

    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Graph
plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.grid(True)
plt.show()
```

```python
# Clustering with Original Data using K-Means

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

#Find Optimal Number of Clusters using Elbow Method

# Elbow Method
wcss = []

for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=42
    )

    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Graph
plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.grid(True)
plt.show()

# Apply K-Means Clustering

# Apply KMeans
kmeans = KMeans(
    n_clusters=3,
    init='k-means++',
    random_state=42
)

# Fit and predict clusters
clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to dataset
df['Cluster'] = clusters

print(df.head())

#Visualize Clustering Results

# Scatter plot

plt.figure(figsize=(8,6))

plt.scatter(
    X_scaled.iloc[:,0],
    X_scaled.iloc[:,1],
    c=clusters
)

# Plot centroids
plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    s=200,
    marker='X'
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering")

plt.grid(True)
plt.show()

# Evaluate Clustering Performance

# Silhouette Score
sil_score = silhouette_score(X_scaled, clusters)

print("Silhouette Score:", sil_score)

# Davies–Bouldin Index
# Davies-Bouldin Score
db_score = davies_bouldin_score(X_scaled, clusters)

print("Davies-Bouldin Score:", db_score)
```

```python
#Apply K-Means on PCA Data
# Apply KMeans on PCA-transformed data

kmeans_pca = KMeans(
    n_clusters=3,
    init='k-means++',
    random_state=42
)

# Predict clusters
pca_clusters = kmeans_pca.fit_predict(X_pca_final)

# Add cluster labels
pca_df['Cluster'] = pca_clusters

print(pca_df.head())

# Visualize PCA clusters

plt.figure(figsize=(8,6))

plt.scatter(
    X_pca_final[:,0],
    X_pca_final[:,1],
    c=pca_clusters
)

# Plot centroids
plt.scatter(
    kmeans_pca.cluster_centers_[:,0],
    kmeans_pca.cluster_centers_[:,1],
    s=200,
    marker='X'
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering on PCA Data")

plt.grid(True)
plt.show()

# Silhouette score for PCA clustering

silhouette_pca = silhouette_score(
    X_pca_final,
    pca_clusters
)

print("Silhouette Score (PCA Data):", silhouette_pca)

# Davies-Bouldin score for PCA clustering

db_pca = davies_bouldin_score(
    X_pca_final,
    pca_clusters
)

print("Davies-Bouldin Score (PCA Data):", db_pca)

#Compare Original Data vs PCA Data Clustering
# Create comparison table

comparison = pd.DataFrame({
    'Metric': ['Silhouette Score', 'Davies-Bouldin Score'],
    'Original Data': [sil_score, db_score],
    'PCA Data': [silhouette_pca, db_pca]
})

print(comparison)

#Visual Comparison using Bar Chart
# Plot comparison

comparison.set_index('Metric').plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Clustering Performance Comparison")
plt.ylabel("Score")

plt.grid(True)
plt.show()

#Compare Cluster Visualization
plt.figure(figsize=(8,6))

plt.scatter(
    X_scaled.iloc[:,0],
    X_scaled.iloc[:,1],
    c=clusters
)

plt.title("Clusters on Original Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.grid(True)
plt.show()

#plt.figure(figsize=(8,6))

plt.scatter(
    X_pca_final[:,0],
    X_pca_final[:,1],
    c=pca_clusters
)

plt.title("Clusters on PCA-Transformed Data")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.grid(True)
plt.show()

#PCA data cluster
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca_final[:,0],
    X_pca_final[:,1],
    c=pca_clusters
)

plt.title("Clusters on PCA-Transformed Data")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.grid(True)
plt.show()
```

Impact of Dimensionality Reduction on Clustering

PCA transforms data into fewer dimensions while preserving maximum variance.

Transformation formula:

Y=XW

Where:

X = original standardized data
W = eigenvector matrix
Y = PCA-transformed data

Positive Impacts of PCA
A. Reduced Noise
PCA removes less important variance.
Helps clustering algorithms focus on meaningful patterns.
B. Faster Computation
Fewer dimensions reduce processing time.
C. Better Cluster Separation
Redundant features are removed.
Clusters may become more compact.
D. Improved Visualization
PCA enables easy 2D or 3D plotting.
Negative Impacts of PCA
A. Information Loss
Some variance is discarded.
B. Reduced Interpretability
Principal components do not have direct real-world meaning.
C. Important Features May Be Removed
Small-variance features may still contain useful information

# 2. Practical Implications of PCA and Clustering in Data Analysis

---

## Practical Uses of PCA

### A. Dimensionality Reduction

PCA helps reduce large feature sets into fewer meaningful components.

### B. Noise Reduction

Unimportant variations are removed, improving analysis quality.

### C. Faster Machine Learning

Reduced dimensions decrease training time and computational cost.

### D. Data Visualization

High-dimensional datasets can be visualized in 2D or 3D.

### E. Multicollinearity Handling

Highly correlated variables are transformed into independent components.

---

## Practical Uses of Clustering

### A. Customer Segmentation

Businesses group customers based on purchasing behavior.

### B. Image Segmentation

Used in computer vision and image processing.

### C. Fraud Detection

Unusual transactions can form separate clusters.

### D. Document Grouping

Text mining and recommendation systems use clustering.

### E. Medical Diagnosis

Patients with similar symptoms can be grouped together.

# 3. Recommendations Based on the Analysis

---

## When to Use PCA

Use PCA when:

| Scenario                    | Reason                          |
| --------------------------- | ------------------------------- |
| High-dimensional data       | Reduces feature complexity      |
| Highly correlated variables | Removes multicollinearity       |
| Slow model performance      | Improves computation speed      |
| Visualization needed        | Enables 2D/3D plotting          |
| Noisy datasets              | Removes less important variance |

---

## When to Use Clustering Directly on Original Data

Use original data when:

| Scenario                         | Reason                               |
| -------------------------------- | ------------------------------------ |
| Dataset has few features         | Dimensionality reduction unnecessary |
| Interpretability is important    | Original features retain meaning     |
| Features are already independent | PCA may not add much value           |
| Minimal noise exists             | Direct clustering works well         |

---

## When to Combine PCA with Clustering

Use PCA before clustering when:

| Scenario                 | Benefit                |
| ------------------------ | ---------------------- |
| Large datasets           | Faster clustering      |
| Many correlated features | Better cluster quality |
| Visualization required   | Clear cluster plots    |
| Noise affects clustering | Improved separation    |
