```python
import pandas as pd
df = pd.read_csv("advertising.csv")
df.head()
```

```python
X_cont = df.iloc[:,0:4]
```

```python
# stndardization
from sklearn.preprocessing import StandardScaler
SS = StandardScaler()
SS_X = SS.fit_transform(X_cont)
SS_X = pd.DataFrame(SS_X)
SS_X.columns = list(X_cont)
SS_X.head()
```

```python
Y = df["Clicked on Ad"]

X = pd.concat([SS_X,df['Male']],axis=1)
X.head()
```

```python
# data partition
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,train_size=0.7)
```

```python
X_train.shape,X_test.shape,Y_train.shape,Y_test.shape
```

```python
# trianing the model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

model.fit(X_train,Y_train)

Y_pred_train = model.predict(X_train)
Y_pred_test = model.predict(X_test)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score

training_score = accuracy_score(Y_train,Y_pred_train)
print("Training Accuracy score: ", np.round(training_score,2))
test_score = accuracy_score(Y_test,Y_pred_test)
print("Test Accuracy score: ", np.round(test_score,2))
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(solver='liblinear')
```

```python
from sklearn.model_selection import ShuffleSplit, cross_validate
shuffle_split = ShuffleSplit(n_splits=100, test_size=0.2)

cv_results = cross_validate(model, X, Y, cv=shuffle_split, scoring = 'accuracy', return_train_score=True)
```

```python
import numpy as np
print("Training Accuracy score: ", np.round(np.mean(cv_results['train_score']),2))
print("Test Accuracy score: ", np.round(np.mean(cv_results['test_score']),2))
```

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.linear_model import LogisticRegression
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Re-define X and Y to ensure they are available in this cell's scope
X_cont = df.iloc[:,0:4]
SS = StandardScaler()
SS_X = SS.fit_transform(X_cont)
SS_X = pd.DataFrame(SS_X)
SS_X.columns = list(X_cont)
Y = df["Clicked on Ad"]
X = pd.concat([SS_X,df['Male']],axis=1)

model = LogisticRegression(solver='liblinear') # Re-initialize the model
KFold = KFold(n_splits=5)

cv_results = cross_validate(model, X, Y, cv=KFold, scoring = 'accuracy', return_train_score=True)

import numpy as np
print("Training Accuracy score: ", np.round(np.mean(cv_results['train_score']),2))
print("Test Accuracy score: ", np.round(np.mean(cv_results['test_score']),2))
```
