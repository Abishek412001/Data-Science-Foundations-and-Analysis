```python
# Import libraries
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

# Sample dataset (replace with your dataset)
df = pd.read_csv('adult_with_headers.csv')

# Replace ' ?' with NaN for consistent missing value handling
df.replace(' ?', np.nan, inplace=True)

print("Original Data:")
print(df)

# -----------------------------
# 1. HANDLE MISSING VALUES
# -----------------------------

# Separate numerical and categorical columns
numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

# Impute numerical columns with mean
mean_imputer = SimpleImputer(strategy='mean')
df[numeric_cols] = mean_imputer.fit_transform(df[numeric_cols])

# Impute categorical columns with most frequent value
mode_imputer = SimpleImputer(strategy='most_frequent')
df[categorical_cols] = mode_imputer.fit_transform(df[categorical_cols])

df_imputed = df.copy()

print("\nAfter Handling Missing Values:")
print(df_imputed)

# -----------------------------
# 2a. STANDARD SCALING
# -----------------------------

standard_scaler = StandardScaler()
df_standard_scaled = pd.DataFrame(
    standard_scaler.fit_transform(df_imputed[numeric_cols]), # Scale only numerical columns
    columns=numeric_cols
)

print("\nStandard Scaled Data (Numerical only):")
print(df_standard_scaled)

# -----------------------------
# 2b. MIN-MAX SCALING
# -----------------------------

minmax_scaler = MinMaxScaler()
df_minmax_scaled = pd.DataFrame(
    minmax_scaler.fit_transform(df_imputed[numeric_cols]), # Scale only numerical columns
    columns=numeric_cols
)

print("\nMin-Max Scaled Data (Numerical only):")
print(df_minmax_scaled)

SimpleImputer(strategy='mean')   # numerical
SimpleImputer(strategy='median')
SimpleImputer(strategy='most_frequent')  # categorical

# 3. ENCODING TECHNIQUES
# -----------------------------

# Identify categorical columns
categorical_cols = df.select_dtypes(include='object').columns

# Separate columns based on unique values
one_hot_cols = [col for col in categorical_cols if df[col].nunique() < 5]
label_cols = [col for col in categorical_cols if df[col].nunique() >= 5]

print("\nColumns for One-Hot Encoding:", one_hot_cols)
print("Columns for Label Encoding:", label_cols)

# -----------------------------
# 3a. ONE-HOT ENCODING (<5 categories)
# -----------------------------

df_one_hot = pd.get_dummies(df[one_hot_cols], drop_first=True)

# -----------------------------
# 3b. LABEL ENCODING (>5 categories)
# -----------------------------

label_encoder = LabelEncoder()

df_label = df[label_cols].apply(label_encoder.fit_transform)

# -----------------------------
# 4. COMBINE ALL DATA
# -----------------------------

# Drop original categorical columns
df_final = df.drop(columns=categorical_cols)

# Concatenate encoded data
df_final = pd.concat([df_final, df_one_hot, df_label], axis=1)

print("\nFinal Processed Data:")
print(df_final)
```
