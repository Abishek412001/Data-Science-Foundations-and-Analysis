```python
#	Provide a brief interpretation of these statistics.
# Mean: This is the average value for each numerical column. It gives you a general idea of the central value. For example, the Volume has a mean of approximately 5.07, suggesting that on average, about 5 units are sold per transaction.

# Median: This is the middle value when the data for each column is sorted. It's often a better measure of central tendency than the mean when there are outliers, as it's not as affected by extreme values. For Volume, the median is 4, which is close to the mean, suggesting the data isn't heavily skewed.

# Mode: This represents the most frequently occurring value in each column. It tells you which specific value appears most often. If a column has multiple modes, they will all be listed.

# Standard Deviation (std): This measures the amount of variation or dispersion of values around the mean. A small standard deviation indicates that the data points tend to be very close to the mean, while a large standard deviation indicates that the data points are spread out over a wider range. For instance, Total Sales Value has a much larger standard deviation than Volume, indicating a wider range of sales values compared to the number of items sold
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define the filename
file_name = 'sales_data_with_discounts.csv'

# Check if the file exists. If not, prompt the user to upload it.
if not os.path.exists(file_name):
    print(f"File '{file_name}' not found in the current directory. Please upload it.")
    from google.colab import files
    uploaded = files.upload()
    if file_name not in uploaded:
        # If the user didn't upload the expected file, raise an error
        raise FileNotFoundError(f"'{file_name}' was not uploaded. Please ensure you upload the correct file.")

# Load the dataset after ensuring it's available
df = pd.read_csv(file_name)

numerical_cols = df.select_dtypes(include=['number']).columns

for col in numerical_cols:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
```

### Interpretation of Histograms

- **Volume**: The distribution of `Volume` appears to be right-skewed, indicating that most transactions involve a small number of items, with fewer transactions involving a large number of items. There might be some outliers at higher volumes.
- **Avg Price**: The `Avg Price` histogram shows a heavily right-skewed distribution, suggesting that the majority of products sold have lower average prices, while a smaller number of products command very high prices. This skewness is typical for price data.
- **Total Sales Value**: Similar to `Avg Price`, `Total Sales Value` is also right-skewed, implying that most sales transactions generate a relatively lower total value, with a few transactions contributing significantly higher sales values.
- **Discount Rate (%)**: The `Discount Rate (%)` seems to have a more uniform distribution, or perhaps multiple peaks, suggesting various discount strategies might be in place. It doesn't appear as heavily skewed as the price and sales values.
- **Discount Amount**: This column also exhibits a right-skewed distribution, consistent with `Total Sales Value`. Higher discount amounts are less frequent.
- **Net Sales Value**: As expected, `Net Sales Value` mirrors `Total Sales Value` with a right-skewed distribution, indicating that the majority of net sales are lower, and a few transactions generate very high net sales.

### Box Plots for Numerical Columns

Next, let's use box plots to visualize the distribution of numerical variables, which are excellent for identifying outliers, quartiles, and the spread of the data.

```python
numerical_cols = df.select_dtypes(include=['number']).columns

for col in numerical_cols:
    plt.figure(figsize=(8, 5))
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
```

### Interpretation of Box Plots

- **Volume**: The box plot for `Volume` shows a concentration of data at lower values, confirming the right-skewness observed in the histogram. There are several outliers at higher volumes, indicating some transactions with unusually large quantities.
- **Avg Price**: The `Avg Price` box plot is highly concentrated at the lower end, with a very long upper whisker and numerous individual data points extending far above the main body of the data. This clearly indicates a strong right-skewness and the presence of many high-priced outlier products.
- **Total Sales Value**: Similar to `Avg Price`, `Total Sales Value` also shows a highly skewed distribution with many outliers on the higher end, indicating transactions that generated significantly more revenue than the typical sale.
- **Discount Rate (%)**: The `Discount Rate (%)` box plot appears more spread out than the price-related variables, but still shows some outliers on both the lower and higher ends, suggesting a range of discount strategies. The median is visible, and the box itself shows where the central 50% of discount rates fall.
- **Discount Amount**: This plot mirrors the `Total Sales Value` and `Avg Price` with a strong right-skewness and numerous high-value outliers, indicating a few transactions involved exceptionally large discount amounts.
- **Net Sales Value**: As expected, `Net Sales Value` is similar to `Total Sales Value`, displaying a right-skewed distribution with a substantial number of high-value outliers, representing transactions with very high net sales.

### Bar Chart Analysis for Categorical Columns

Now, let's analyze the categorical variables to understand the frequency distribution of their unique values. This will help us identify the most common categories and any imbalances.

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(y=df[col], order=df[col].value_counts().index, palette='viridis')
    plt.title(f'Frequency of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.grid(axis='x', linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
```

### Interpretation of Bar Charts

- **Date**: The bar chart for `Date` would show the frequency of transactions on each date. If the dataset covers a short period, it might indicate daily transaction volumes. If it spans a longer period, it could reveal trends over time, such as days with higher or lower activity.
- **Day**: The `Day` column's bar chart will illustrate the distribution of transactions across days of the week. This can reveal if certain days (e.g., weekends) have higher sales activity or if there are specific patterns throughout the week.
- **SKU**: The `SKU` (Stock Keeping Unit) bar chart will highlight the most frequently sold products. A highly skewed distribution would indicate that a few products dominate sales, while a more uniform distribution would suggest a wider variety of popular items.
- **City**: The bar chart for `City` will show the distribution of sales across different cities. This can help identify the most active sales regions and assess geographical performance.
